# **Exploratory Data Analysis (EDA)**

---

## Introduction

This exploratory data analysis (EDA) examines customer purchasing behavior, product performance, and sales trends using a cleaned e-commerce dataset analyzed with Python Pandas. The analysis aims to generate insights that support better business and sales decision-making.

### Objectives

- Analyse customer order value distribution and purchasing patterns.
- Segment customers by geographical information to identify regional purchasing differences.
- Examine sales performance across different products.
- Compare the number of orders with items added to carts to identify purchasing trends.
- Investigate the influence of discounts on customer purchases.
- Assess product popularity based on the number of orders containing each product.
- Identify and visualise trends in purchases and sales over time.

---


##### Load required Libraries

```python
import pandas as pd
import numpy as np
from datetime import datetime

# Visualization Libraries
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

# Settings
sns.set_style('dark')
plt.rcParams['figure.figsize'] = (10,6)

import warnings
```

#### Load Datset

```python
file_name ='Cleaned_Dataset.xlsx'
df = pd.read_excel(file_name)
```

```python
df.head()
```

----

**Helper Columns for Trend Analysis:**

```python
df['Month'] = df['Date'].dt.month
df['MonthName'] = df['Date'].dt.strftime('%B')
df['DayOfWeek'] = df['Date'].dt.day_name()
df['Year'] = df['Date'].dt.year
```

```python
# Overal Key Performing Indicators
total_orders = df['OrderID'].nunique()
total_revenue = df['TotalPrice'].sum()
total_customers = df['CustomerID'].nunique()
order_value = total_revenue / total_orders 
purchase_rate = df['Quantity'].sum() / df['ItemsInCart'].sum() 
print(f'Total Customers: {total_customers}')
print(f'Total Orders: {total_orders}')
print(f'Total Revenue: ${total_revenue:,.2f}')
print(f'Average Order Value: ${order_value:,.2f}')
print(f'Purchase Rate: {purchase_rate:.2%}')
```

---

##### **How revenue has evolved Overtime**

**1. Revenue By Months**

```python
# Define Correct Month Order
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

# Convert MonthName to Ordered Category
df['MonthName'] = pd.Categorical(
    df['MonthName'],
    categories=month_order,
    ordered=True
)
```

```python
# Visualize Revenue Trends
revenue_trend = df.groupby('MonthName', observed=False)['TotalPrice'].sum().round(2).reset_index()


plt.figure(figsize=(10,5))
    
plt.plot(revenue_trend['MonthName'],
         revenue_trend['TotalPrice'],marker='o')
plt.title('Monthly Sales')
plt.ylabel('Total Sales')
plt.show()
```

----

**Weekly Order Performance**

```python
days_order = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']

# Convert Weekday Name In Categorical Order
df['DayOfWeek'] = pd.Categorical(
    df['DayOfWeek'],
    categories= days_order,
    ordered=True
)
```

```python
# Group Orders by weekday
weekly_orders = df.groupby('DayOfWeek', observed=False)['OrderID'].nunique().reset_index().sort_values('DayOfWeek')


# Visualize Orders by weekday
plt.figure(figsize=(8,5))
plt.bar(weekly_orders['DayOfWeek'],
            weekly_orders['OrderID'])

plt.title('Weekly Order Performance')
plt.xlabel('Weekday')
plt.ylabel('Order Count')
plt.show()
```

##### **Purchase Segmentation by Location**

**1.Total Sales by Location**

```python
location_sales = df.groupby('ShippingAddress')['TotalPrice'].sum().round(2).sort_values(ascending=False)
```

**2.Visualize Sales by Location**

```python
location_sales.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Cities by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Cities')
plt.show()
```

----

##### **Product Purchase Analysis:**

**1. Sales Across Product Categories**

```python
# Group sales by product category
product_sales = df.groupby('Product')['TotalPrice'].sum().round(2).sort_values(ascending=False)
```

```python
# Visualize product performance
product_sales.head(10).sort_values().plot(kind='barh')

plt.title('Top 10 Products by Sales')
plt.xlabel('Sales Revenue')  
plt.ylabel('Products')
plt.show()
```

**2.Product Order Quantity**

```python
product_popularity = df.groupby('Product').agg(
    {'Quantity': 'sum', 
     'OrderID': 'nunique'}
    ).sort_values(
        by='Quantity', ascending=False)

product_popularity.rename(
    columns={'Quantity':'TotalQuantity',
             'OrderID':'NumOrders'}, inplace=True)

product_popularity
```

**3. Visualize product popularity:**

```python
product_popularity = df.groupby('Product')['OrderID'].nunique().sort_values(ascending=False)

product_popularity.sort_values().plot(kind='barh')

plt.title('Most Frequently Ordered Products')
plt.xlabel('Product')
plt.ylabel('Number of Orders')
plt.show()
```

#### **Orders vs Items added to Cart**

This helps identify:

```python
# Compare Metrics
cart_size = df.groupby('OrderID').agg({'ItemsInCart':'sum',
                                       'Quantity':'sum'}).reset_index()

cart_size.head(10)
```

----

##### **3. How Discounts Influence Purchases**


```python
# Identify Discount Usage
df['UsedCoupon'] = np.where(df['CouponCode'] == 'No Coupon',
                            'No Discount', 
                            'Discount Applied')

# Average Order value with Coupon
discount_anlysis = df.groupby('UsedCoupon')['TotalPrice'].sum()

# Visualize discount impact
plt.pie(
    discount_anlysis,
    labels = discount_anlysis.index,
    autopct='%1.0f%%',
    startangle = 90,
    wedgeprops={'width':0.6}
)
plt.title('Sales Distribution: Discount vs No Discount')
plt.show()
```
